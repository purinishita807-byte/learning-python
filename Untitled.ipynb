```python
dna_sequence = "ATGAAACGCATTAGC"

g_count = dna_sequence.count("G")
c_count = dna_sequence.count("C")

gc_count = g_count + c_count
print("GC Count:", gc_count)
```

```python
dna_sequence = "ATGAAACGCATTAGC"

# 1. (Index 0)
first_base = dna_sequence[0]
print("First Base:", first_base)

# 2. (Start Codon)
first_codon = dna_sequence[0:3]
print("First Codon:", first_codon)

# 3. (Second Codon)
second_codon = dna_sequence[3:6]
print("Second Codon:", second_codon)
```
