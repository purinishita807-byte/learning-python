```python
# 1. 'calculate_gc'
def calculate_gc(dna_sequence):
    gc_count = 0
    for base in dna_sequence:
        if base == "G" or base == "C":
           gc_count = gc_count + 1
    return gc_count

# 2.(Function Call)
patient_1 = "ATGCGC"
patient_2 = "AATTGGCCAA"

# patient  1
result_1 = calculate_gc(patient_1)
# patient  2
result_2 = calculate_gc(patient_2)

print("Patient 1 GC Count:", result_1)
print("Patient 2 GC Count:", result_2)
```
