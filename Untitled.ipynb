```python
# 1.(15bp)
dna_sequence = "ATGAAACGCATTAGC"
print("1. Original DNA:", dna_sequence)

# 2.(Length)
sequence_len = len(dna_sequence)
print("2. Sequence Length:", sequence_len)

# 3. .lower() method
lowercase_dna = dna_sequence.lower()
print("3. Lowercase DNA:", lowercase_dna)

# 4. ADVANCED BIO-LOGIC: DNA-Rna (Transcription)
# .replace() method
rna_sequence = dna_sequence.replace("T", "U")
print("4. Transcribed RNA:", rna_sequence)
```
