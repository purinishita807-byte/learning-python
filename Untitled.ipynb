```python
# 1. Function define
def transcribe_dna_to_rna(dna_sequence):
# Use .replace()  T to U 'rna'
    rna = dna_sequence.replace("T", "U")
    return rna

# 2. Function test
my_dna = "ATGAAACGCATTAGC"
my_rna = transcribe_dna_to_rna(my_dna)
print("My Transcribed RNA is:", my_rna)
```
