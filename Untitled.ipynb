```python
codon = "AUG"

# 1. Check if it is a Start Codon
if codon == "AUG":
    print("🔬 This is a START Codon. Begin Translation!")

# 2. Check if it is a Stop Codon
elif codon == "UAA" or codon == "UAG" or codon == "UGA":
    print("🛑 This is a STOP Codon. Stop Translation!")

# 3. If it is neither Start nor Stop
else:
    print("🧬 This is a normal Amino Acid Codon.")
```
