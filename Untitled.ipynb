```python
# 1. (Class)
class Virus:
    # __init__ use
    def __init__(self, name, sequence):
        self.name = name
        self.sequence = sequence

    # GC content in-built function
    def get_len(self):
        return len(self.sequence)

# 2. (Object)
covid = Virus("COVID-19", "ATGAAACGC")
ebola = Virus("Ebola", "AATTTGGCCAAA")

# 3. Data excess
print(f"Virus Name: {covid.name}, Length: {covid.get_len()}bp")
print(f"Virus Name: {ebola.name}, Length: {ebola.get_len()}bp")
```
