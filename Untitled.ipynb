```python
gc_percentage = 35.5
if gc_percentage >= 50:
    print("👍 Good Primer for PCR!")
else:
    print("❌ Bad Primer. DNA binding will be weak.")
```

```python
dna_chain = "ATGAAUGC"

# Loop
for base in dna_chain:
    if base == "U":
       print("⚠️ Warning: RNA contamination or Mutation found ('U')!")
    else:
       print(f"Base {base} is normal.")
```

```python
dna = "ATGCGC"
gc_count = 0

for base in dna:
    if base == "G" or base == "C":
       gc_count = gc_count + 1
print("Total GC Content is:", gc_count)
```
