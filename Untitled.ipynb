```python
# 1.(Genetic Code Table)
genetic_code = {
"AUG": "Methionine",
"UUU": "Phenylalanine",
"CGC": "Arginine",
"CGG": "Arginine"
}
# 2.Specific Amino acid
codon = "AUG"
amino_acid = genetic_code[codon]
print(f"Codon {codon} codes for: {amino_acid}")
# output: Codon AUG codes for: Methionine

# 3.  Stop Codon Add
genetic_code["UAA"] = "STOP"
print("Updated Dictionary:", genetic_code)
```
