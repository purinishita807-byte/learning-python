```python
# 1.make list
patient_sequences = ["ATGAAACGC", "ATGAAACGG", "ATGAAACGA"]
print("1. Initial Database:", patient_sequences)

# 2.sequence print
first_patient = patient_sequences[0]
print("2. First Patient DNA:", first_patient)

# 3. .append() use
new_patient = "ATGAAACGT"
patient_sequences.append(new_patient)
print("3. Updated Database (After Append):", patient_sequences)

# 4.len() use
total_patients = len(patient_sequences)
print("4. Total Patients in Database:", total_patients)
```
